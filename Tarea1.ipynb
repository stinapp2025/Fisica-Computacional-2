```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
def f(x):
    return x**3
def f_d(x,h):
    return (f(x+h) - f(x-h))/(2*h)
h = np.geomspace(1e-20,0.1,200)
x = 1
epsilon_star = 2**-52
E_abs = (h**2/6) * np.abs(6) +(epsilon_star/h) * np.abs(f(x))
#print(E_abs)
print(f_d(x,h))
E_num = np.abs(f_d(x,h) - 3)
```

```python
plt.xscale("log")
plt.xlabel("h")
plt.yscale("log")
plt.ylabel("Error absoluto")
plt.title("Gráfico del Error absoluto de una derivada en comparación con un incremento de h")
plt.scatter(h,E_num)
plt.plot(h,E_abs)
```
